# Ejercicio: Definición de Skills para Agentes de IA
Basándose en los conceptos de plataformas como AI Hero, definir programáticamente 4 *skills* (habilidades o roles) para un asistente o agente de Inteligencia Artificial. 

Las habilidades a definir son:
1. **Documentador de Tickets:** Para documentar toda la descripción de un trabajo por hacer.
2. **Implementador de Código:** Para la generación de código y pruebas.
3. **Explicador Técnico:** Para generar un HTML que explique en detalle el trabajo realizado.
4. **Capturador de Requerimientos:** Para leer y extraer requerimientos de documentos provistos.

En este notebook, modelaremos estos *skills* utilizando diccionarios de Python que representan la configuración (Nombre, Descripción y *System Prompt*) que se enviaría a una API de Modelos de Lenguaje (como OpenAI, Claude o Gemini).

```python
# Estructura base para almacenar nuestros Skills
class AISkill:
    def __init__(self, nombre, descripcion, system_prompt):
        self.nombre = nombre
        self.descripcion = descripcion
        self.system_prompt = system_prompt
        
    def __str__(self):
        return f"--- SKILL: {self.nombre} ---\nDescripción: {self.descripcion}\nPrompt:\n{self.system_prompt}\n"

# Diccionario para almacenar el catálogo de skills
catalogo_skills = {}
```

## Skill 1: Documentación de un Ticket de Trabajo
Este skill se encargará de recibir una idea o solicitud informal y transformarla en un ticket de trabajo estructurado (estilo Jira o Trello).

```python
catalogo_skills['documentador_tickets'] = AISkill(
    nombre="Documentador de Tickets de Trabajo",
    descripcion="Toma una descripción general de una tarea y genera un ticket estructurado detallando el trabajo por hacer.",
    system_prompt="""Eres un Product Manager y Scrum Master experto. 
Tu tarea es recibir la descripción de una necesidad del usuario y transformarla en un ticket de trabajo formal.
Debes incluir obligatoriamente:
- Título claro y conciso.
- Descripción detallada del trabajo por hacer.
- Criterios de Aceptación (Acceptance Criteria) numerados.
- Tareas técnicas sugeridas (Sub-tasks).
Devuelve únicamente la estructura del ticket en formato Markdown."""
)

print(catalogo_skills['documentador_tickets'])
```

## Skill 2: Implementación y Pruebas
Este skill actuará como un desarrollador de software. Tomará el ticket generado en el paso anterior y escribirá el código fuente y sus pruebas unitarias.

```python
catalogo_skills['implementador_codigo'] = AISkill(
    nombre="Implementador de Código y Pruebas",
    descripcion="Genera el código fuente y las pruebas unitarias basándose en un ticket de trabajo.",
    system_prompt="""Eres un Desarrollador de Software Senior. 
Tu tarea es leer un ticket de trabajo estructurado y generar el código necesario para su implementación.
Reglas:
1. Escribe código limpio, eficiente y bien comentado en Python.
2. Implementa siempre un conjunto de pruebas unitarias usando la librería `unittest` o `pytest` para garantizar que el código cumple con los Criterios de Aceptación.
3. Separa el bloque de código de implementación del bloque de pruebas."""
)

print(catalogo_skills['implementador_codigo'])
```

## Skill 3: Explicación y Reporte en HTML
Este skill toma el trabajo técnico realizado (código y pruebas) y lo traduce a un formato visual y fácil de leer (HTML) para presentar el reporte final a los interesados.

```python
catalogo_skills['explicador_html'] = AISkill(
    nombre="Explicador de Trabajo en HTML",
    descripcion="Genera un documento HTML detallado que explica el trabajo realizado y el código implementado.",
    system_prompt="""Eres un Comunicador Técnico (Technical Writer). 
Tu tarea es recibir un código fuente y la descripción de su implementación, y generar un archivo HTML5 válido y bien estructurado que explique en detalle el trabajo realizado.
Reglas:
- Usa etiquetas semánticas (<h1>, <section>, <code>, <pre>).
- Incluye un resumen ejecutivo, explicación de la arquitectura/decisiones técnicas, y un manual de uso básico.
- El HTML debe contener estilos CSS integrados básicos para que sea estéticamente agradable.
- Devuelve únicamente código HTML, sin texto introductorio."""
)

print(catalogo_skills['explicador_html'])
```

## Skill 4: Capturador de Requerimientos
Este skill es fundamental para las etapas iniciales de un proyecto. Se encarga de analizar documentos de texto en crudo para extraer las necesidades del negocio.

```python
catalogo_skills['capturador_requerimientos'] = AISkill(
    nombre="Capturador de Requerimientos de Documentos",
    descripcion="Lee documentos provistos y extrae los requerimientos funcionales y no funcionales.",
    system_prompt="""Eres un Analista de Requerimientos de Software. 
Tu tarea es leer detenidamente los documentos o textos provistos por el usuario y extraer todos los requerimientos clave.
Debes categorizar tu respuesta de la siguiente manera:
1. Requerimientos Funcionales (Lo que el sistema debe hacer).
2. Requerimientos No Funcionales (Rendimiento, seguridad, diseño, etc.).
3. Reglas de Negocio identificadas.
Asegúrate de ser exhaustivo y capturar cada detalle importante mencionado en el documento provisto."""
)

print(catalogo_skills['capturador_requerimientos'])
```

## Simulación del Flujo de Trabajo (Pipeline)
A continuación, simulamos cómo estos *skills* interactuarían en un flujo de trabajo real, orquestando las llamadas a las funciones.

```python
def simular_agente_ia(skill_id, input_data):
    """
    Función simulada que representa la llamada a una API de IA utilizando el skill definido.
    En un entorno real, aquí se usaría openai.ChatCompletion.create(...)
    """
    skill = catalogo_skills.get(skill_id)
    if not skill:
        return "Skill no encontrado."
    
    print(f"Ejecutando Skill: {skill.nombre}...")
    print(f"[{skill.nombre}] procesando los siguientes datos:\n{input_data[:50]}...\n")
    
    # Retorno simulado para fines académicos
    return f"[Resultado simulado generado por el skill {skill_id} basado en el input proporcionado]"

# Flujo de ejemplo:
# 1. Capturamos requerimientos de un texto en crudo
texto_crudo = "Necesitamos una calculadora que sume y reste, y debe responder en menos de 1 segundo."
requerimientos = simular_agente_ia('capturador_requerimientos', texto_crudo)

# 2. Documentamos un ticket basado en esos requerimientos
ticket = simular_agente_ia('documentador_tickets', requerimientos)

# 3. Implementamos el código basado en el ticket
codigo = simular_agente_ia('implementador_codigo', ticket)

# 4. Generamos la explicación HTML
html = simular_agente_ia('explicador_html', codigo)

print("--- FLUJO COMPLETADO ---")
```
